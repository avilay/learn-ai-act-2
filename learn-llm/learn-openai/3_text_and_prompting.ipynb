```python
from openai import OpenAI
from utils import LLM
```

```python
client = OpenAI()
```

```python
resp = client.chat.completions.create(
    model=LLM.PRE_FAST_MINI,
    messages=[
        {
            "role": "user",
            "content": "Write a one-sentence bedtime story about a unicorn",
        }
    ],
)
print(resp.choices[0].message.content)
```

```python
resp = client.chat.completions.create(
    model=LLM.PRE_FAST_MINI,
    messages=[
        {"role": "developer", "content": "Talk like a pirate"},
        {"role": "user", "content": "Are semicolons optional in Javascript?"},
    ],
)
```

```python
print(resp.choices[0].message.content)
```

```python
# Check if the model is still talking in pirate
resp = client.chat.completions.create(
    model=LLM.PRE_NANO,
    messages=[{"role": "user", "content": "What are ways to go to bed on time?"}],
)
```

```python
print(resp.choices[0].message.content)
```

### Roles

**DEVELOPER**: `developer` messages are instructions provided by the application developer, weighted ahead of user messages. This was previously called the `system` prompt.

**USER**: `user` messages are instructions provided by the end user of the application, weighted lower than `developer` messages.

**ASSISTANT**: Messages generated by the model have the `assistant` role.

### Models
Which model to use?

  * **Reasoning models** generate an internal chain of thought to analyze the input prompt, and excel at understanding complex tasks and multi-step planning. They are also generally slower and more expensive to use than GPT models.

  * **GPT models** are fast, cost-efficient, and highly intelligent, but benefit from more explicit instructions around how to accomplish tasks.

  * **Large and small (min) models** offer trade-offs for speed, cost, and intelligence. Large models are more effective at understanding prompts and solving problems across domains, while small models are generally faster and cheaper to use. Small models like GPT-4o can also be trained to excel at a specific task through fine-tuning and distillation of results from larger models.

  When in doubt, `gpt-4o` offers a solid combination of intelligence, speed, and cost effectiveness.
