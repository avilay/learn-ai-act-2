# Completion API
This is the older API but I am going to base my learning on this because other providers have copied this API. The basic calling pattern is simple - I call the `create` function with at least two params - `model` and `messages`. The `model` is just a string of whatever model I want to use. `messages` is a list of messages with each message being a dict with at least two keys - `role` and `content`. The `role` can be "developer", "user", or "assistant" as explained below. The `content` can be just a string of the prompt. Or it can be a list of typed dicts. E.g., here are some example messages.

A very simple message -
```python
{
    "role": "developer",
    "content": "You are a math tutor."
}
```

A message with an image url -
```python
{
    "role": "user",
    "content": [
        {
            "type": "text",
            "text": "What teams are playing in this image?"
        },
        {
            "type": "image_url",
            "image_url": {
                "url": "https://upload.wikimedia.org/wikipedia/commons/3/3b/LeBron_James_Layup_%28Cleveland_vs_Brooklyn_2018%29.jpg"
            },
        },
    ]
}
```

A message with multiple base64 encoded images -
```python
{
    "role": "user",
    "content": [
        {
            "type": "text",
            "text": "A total of how many people are in these images?",
        },
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{b64img1}",
                "detail": "low",
            },
        },
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{b64img2}",
                "detail": "low",
            },
        },
    ],
}
```

A message with a base64 encoded pdf -
```python
{
    "role": "user",
    "content": [
        {
            "type": "file",
            "file": {
                "filename": "draconomicon.pdf",
                "file_data": f"data:application/pdf;base64,{base64_string}",
            }
        },
        {
            "type": "text",
            "text": "What is the first dragon in the book?",
        }
    ],
}
```
